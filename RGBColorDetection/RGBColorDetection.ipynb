```python
from ultralytics import YOLO
import numpy as np
import cv2
```

```python
model = YOLO('yolov8n.pt')
```

```python
video_path = './traffic_light.mp4'
output_path = './op_sample08.mp4'
video_cap = cv2.VideoCapture(video_path)
ret, frame = video_cap.read()
H, W ,_ = frame.shape
output = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'MP4V'), int(video_cap.get(cv2.CAP_PROP_FPS)),(W,H))
thresh = 0.5
green, yellow, red = (0,255,0),(30,192,243),(0,0,255)
while ret:
    if not ret:
        print('Empty frame')
        continue
    results = model.track(frame, stream=True)
    for detection in results:
        classes_names = detection.names
        for bbox in detection.boxes:
            if bbox.conf[0] > thresh:
                [x1,y1,x2,y2] = bbox.xyxy[0]
                x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
                cls = int(bbox.cls[0])
                class_name = classes_names[cls]
                if class_name == 'traffic light':
                    g, r = frame[y1:y2, x1:x2, 1], frame[y1:y2, x1:x2, 2]
                    g_mean, r_mean = np.mean(g), np.mean(r)
                    
                    if r_mean > g_mean:
                        color = red
                        text = 'Red Light - STOP'
                    elif g_mean > r_mean:
                        color = green
                        text = 'Green Light - GO'
                    cv2.rectangle(frame, (x1,y1), (x2, y2), color, 3)
                    cv2.putText(frame,f'{class_name}',(x1,y1), cv2.FONT_HERSHEY_SIMPLEX, 2, (150,150,250) ,4)
                    cv2.putText(frame,text,(100, 50), cv2.FONT_HERSHEY_SIMPLEX, 2, color ,4)
    output.write(frame)
    ret, frame=video_cap.read()
video_cap.release()
output.release()
cv2.destroyAllWindows()
```
